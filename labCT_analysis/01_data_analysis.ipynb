# In-Situ (pre/post-operando) imaging of CO2ELY

## Membrane swelling without CL

```python
from skimage import io
import numpy as np
import os

wet_path = '/mnt/nas_nanotomData/CT_Data_PSI/FR54/2023_COELY_postmortem/Z_series/Z_1_wet_2/segmented_BPM.tif'
dry_path = '/mnt/nas_nanotomData/CT_Data_PSI/FR54/2023_COELY_postmortem/Z_series/Z_1_dry_1/segmented.tif'
```

```python
import matplotlib.pyplot as plt
```

```python
from matplotlib.colors import Normalize
```

```python
dry = io.imread(dry_path)
```

```python
wet = io.imread(wet_path)
```

```python
dry.shape
```

```python
plt.imshow(dry[300,...])
```

```python
dry_aem = (dry==1).sum(axis = 2)
dry_cem = (dry==0).sum(axis = 2)
wet_aem = (wet==1).sum(axis = 2)
wet_cem = (wet==0).sum(axis = 2)
```

```python
fig, axs = plt.subplots(1,4, figsize=(10,10))
vmin=0
vmax=35

norm = Normalize(vmin=vmin, vmax=vmax)


for ax in axs:
    ax.axis('off')
axs[0].imshow(dry_aem, vmin=vmin, vmax=vmax)
axs[0].set_title('dry aem')
axs[1].imshow(wet_aem, vmin=vmin, vmax=vmax)
axs[1].set_title('wet aem')
axs[2].imshow(dry_cem, vmin=vmin, vmax=vmax)
axs[2].set_title('dry cem')
axs[3].imshow(wet_cem, vmin=vmin, vmax=vmax)
axs[3].set_title('wet cem')

# fig.colorbar(plt.cm.ScalarMappable(norm=norm, cmap='viridis'),
             # ax=ax, label='thickness [px]')
```

```python
hist_dry_aem = np.histogram(dry_aem, bins=np.arange(40), density = True)
hist_wet_aem = np.histogram(wet_aem, bins=np.arange(40), density = True)
hist_dry_cem = np.histogram(dry_cem, bins=np.arange(40), density = True)
hist_wet_cem = np.histogram(wet_cem, bins=np.arange(40), density = True)
```

```python
plt.plot(hist_dry_aem[1][1:]*3, hist_dry_aem[0], 'k', label='dry aem')
plt.plot(hist_wet_aem[1][1:]*3, hist_wet_aem[0], 'k--', label='wet aem')

plt.plot(hist_dry_cem[1][1:]*3, hist_dry_cem[0], 'r', label='dry cem')
plt.plot(hist_wet_cem[1][1:]*3, hist_wet_cem[0], 'r--', label='wet cem')


plt.xlabel('thickness [um]')
plt.ylabel('frequency [-]')
plt.legend()
```

```python
wet_path = '/mnt/nas_nanotomData/CT_Data_PSI/FR54/2023_COELY_postmortem/B_series/B_6/B_6_wet_2_BPM_segmented.tif'
dry_path = '/mnt/nas_nanotomData/CT_Data_PSI/FR54/2023_COELY_postmortem/B_series/B_6/B_6_dry_BPM_segmented.tif'
```
