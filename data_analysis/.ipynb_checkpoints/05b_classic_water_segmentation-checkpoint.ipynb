```python
from skimage import filters
```

```python
def Gaussian_Blur_4D(da, sigma):
    # TODO: check on boundary mode
    deptharray = np.ones(da.ndim)+4*sigma
    deptharray = tuple(np.min([deptharray, da.shape], axis=0))
    G = da.map_overlap(filters.gaussian, depth=deptharray, boundary='nearest', sigma = sigma)
    return G
```

```python
def plot_im_histogram(im):
    hist = np.histogram(im, bins=100)
    plt.plot(hist[1][1:],hist[0])
```

```python
G = Gaussian_Blur_4D(da, 2)
# G = filters.gaussian(da, sigma = 2)
```

```python
valid_lim = test2.shape[1]-40
print(valid_lim)

valid_limit_dict = {}
valid_limit_dict['4'] = 80
valid_limit_dict['5'] = 60
valid_limit_dict['6'] = 70
valid_limit_dict['3II'] = 85
valid_limit_dict['5II'] = 85
valid_limit_dict['4II'] = 75
valid_limit_dict['8x'] = 60
valid_limit_dict['3III'] = 90
valid_limit_dict['7x'] = 90
water_th = 0.004
```

```python
fig, axs = plt.subplots(1,2, figsize=(19,10))
# axs[0].imshow(np.bitwise_and((test2>0.004),(test2<0.008)))#*test2+test2)
axs[0].imshow((test2>water_th))#*test2+test2)
axs[1].imshow(test2, vmin=0, vmax=0.01)
axs[0].vlines(valid_lim, 0, test2.shape[0]-1)
axs[1].vlines(valid_lim, 0, test2.shape[0]-1)
```

```python
water = np.bitwise_and((ref_min_gauss>water_th),(ref_min_gauss<water_th*2))
```

```python
watervol = np.count_nonzero(water[:,:valid_lim,:], axis=(0,1,2))
watervol2 = np.count_nonzero(water, axis=(0,1,2))
```

```python
plt.figure(figsize=(40,12))
plt.plot((t_utc-t_utc[3])/60,watervol,'x')
plt.plot((t_utc-t_utc[3])/60,watervol2,'x')
```

```python
seg_data = xr.Dataset({'water': (['x','y','z','t_utc'], water*255),
                       'water_volume': ('t_utc', watervol),
                       'time': ('t_utc', time)
                      },
                      coords = {'x': np.arange(water.shape[0]),
                                'y': np.arange(water.shape[1]),
                                'z': np.arange(water.shape[2]),
                               't_utc': t_utc})
seg_data.attrs = data.attrs.copy()
seg_data.attrs['water_segmentation_gitsha'] = git_sha
seg_data.attrs['abcdef_cathode_cropping'] = (a,b,c,e,f)
seg_data.attrs['a1b1c1d1e1f1_cathode_GDL_cropping'] = (a1,b1,c1,e1,f1)
seg_data.attrs['comment'] = 'many artifacts: ring, displacement'
```

## Part 2 check results

```python
watervol_dict = {}

for file in os.listdir(segpath):
    segdata = xr.open_dataset(os.path.join(segpath, file))
    name = segdata.attrs['name']
    print(name)
    water = segdata['water'].data
    t_utc = segdata['t_utc'].data
    print('count pixels')
    watervol = np.count_nonzero(water, axis=(0,1,2))
    segdata.close()
    watervol_dict[name] = (t_utc, watervol)
```
